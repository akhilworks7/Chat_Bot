```python
import os
import time
import pymupdf
import ocrmypdf


# ---------------------------------------------------------
# 1. Check whether PDF is searchable
# ---------------------------------------------------------

def is_searchable_pdf(pdf_path, min_chars_per_page=20):
    """
    Check whether a PDF contains a searchable text layer.

    Returns:
        True  -> PDF is searchable
        False -> PDF is scanned/non-searchable
    """

    doc = pymupdf.open(pdf_path)

    total_pages = len(doc)
    searchable_pages = 0

    for page in doc:
        text = page.get_text().strip()

        if len(text) >= min_chars_per_page:
            searchable_pages += 1

    doc.close()

    print(f"Total pages      : {total_pages}")
    print(f"Searchable pages : {searchable_pages}")

    # If more than 50% of pages have text,
    # consider the document searchable.
    if total_pages > 0:
        ratio = searchable_pages / total_pages
    else:
        ratio = 0

    print(f"Text page ratio  : {ratio:.2%}")

    return ratio >= 0.5


# ---------------------------------------------------------
# 2. Run OCRmyPDF
# ---------------------------------------------------------

def run_ocr(pdf_path):
    """
    Run OCRmyPDF on a scanned/non-searchable PDF.
    """

    base_name = os.path.splitext(pdf_path)[0]

    ocr_pdf = base_name + "_ocr.pdf"

    print("\n🔍 Starting OCRmyPDF...")
    print(f"Input  : {pdf_path}")
    print(f"Output : {ocr_pdf}")

    start_time = time.time()

    ocrmypdf.ocr(
        pdf_path,
        ocr_pdf,

        # Do not OCR pages that already have text
        skip_text=True,

        # Disable unnecessary processing
        deskew=False,
        clean=False,
        optimize=0,

        # OCR language
        language="eng"
    )

    elapsed = time.time() - start_time

    print(f"\n✅ OCR completed")
    print(f"⏱️ OCR time: {elapsed:.2f} seconds")

    return ocr_pdf


# ---------------------------------------------------------
# 3. Prepare PDF
# ---------------------------------------------------------

def prepare_pdf(pdf_path):
    """
    Decide whether to use the original PDF
    or OCR the PDF first.
    """

    print("=" * 60)
    print("PDF PROCESSING")
    print("=" * 60)

    if not os.path.exists(pdf_path):
        raise FileNotFoundError(f"PDF not found: {pdf_path}")

    print(f"Input PDF: {pdf_path}")

    # Check searchable
    searchable = is_searchable_pdf(pdf_path)

    print()

    if searchable:

        print("✅ PDF is searchable")
        print("➡️ Sending original PDF to OpenDataLoader")

        return pdf_path

    else:

        print("❌ PDF is scanned / non-searchable")
        print("➡️ Sending PDF to OCRmyPDF")

        ocr_pdf = run_ocr(pdf_path)

        print("➡️ OCR'd PDF is ready for OpenDataLoader")

        return ocr_pdf


# ---------------------------------------------------------
# 4. Main
# ---------------------------------------------------------

pdf_path = "data/Medical_book.pdf"

processed_pdf = prepare_pdf(pdf_path)

print("\n" + "=" * 60)
print("FINAL PDF")
print("=" * 60)

print(processed_pdf)
```

```python
from pathlib import Path
from langchain_text_splitters import RecursiveCharacterTextSplitter


# ==========================================
# 1. TXT file path
# ==========================================

txt_file = Path(r"output/Medical_book.txt")


# ==========================================
# 2. Read extracted text
# ==========================================

text = txt_file.read_text(encoding="utf-8")

print("Total characters:", len(text))


# ==========================================
# 3. Create Recursive Character Text Splitter
# ==========================================

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=[
        "\n\n",   # Paragraph
        "\n",     # Line
        ". ",     # Sentence
        " ",      # Word
        ""        # Character
    ]
)


# ==========================================
# 4. Create chunks
# ==========================================

chunks = text_splitter.split_text(text)


# ==========================================
# 5. Check chunks
# ==========================================

print("Number of chunks:", len(chunks))


for i, chunk in enumerate(chunks):

    print("\n" + "=" * 80)
    print(f"CHUNK {i + 1}")
    print("=" * 80)

    print(chunk)
```

```python
import torch
from sentence_transformers import SentenceTransformer

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using:", device)

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2",
    device=device
)

embeddings = model.encode(
    chunks,
    batch_size=128 if device == "cuda" else 32,
    normalize_embeddings=True,
    show_progress_bar=True
)

print("Embedding shape:", embeddings.shape)
```

```python
vectors = []

for i, (chunk, embedding) in enumerate(zip(chunks, embeddings)):

    vector = {
        "id": f"chunk_{i}",
        "values": embedding.tolist()
            if hasattr(embedding, "tolist")
            else embedding,
        "metadata": {
            "text": chunk,
            "chunk_id": i,
            "source": "Medical_book.pdf"
        }
    }

    vectors.append(vector)

print(f"Prepared {len(vectors)} vectors")
```

```python
namespace = "documents"
batch_size = 100

for start in range(0, len(vectors), batch_size):

    batch = vectors[start:start + batch_size]

    index.upsert(
        vectors=batch,
        namespace=namespace
    )

    end = min(start + batch_size, len(vectors))

    print(f"Uploaded {end}/{len(vectors)}")
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
def retrieve_documents(query, top_k=5):

    # Create query embedding
    query_embedding = embedding_model.encode(
        query,
        normalize_embeddings=True
    )

    # Search Pinecone
    results = index.query(
        namespace="documents",
        vector=query_embedding.tolist(),
        top_k=top_k,
        include_metadata=True
    )

    documents = []

    for match in results["matches"]:

        documents.append({
            "id": match["id"],
            "score": match["score"],
            "text": match["metadata"].get("text", ""),
            "source": match["metadata"].get("source", ""),
            "page": match["metadata"].get("page", "")
        })

    return documents
```

```python
from groq import Groq

groq_client = Groq(
    api_key=GROQ_API_KEY
)
MODEL_NAME = "openai/gpt-oss-120b"
```

```python
def build_context(documents):

    context_parts = []

    for i, doc in enumerate(documents, start=1):

        context_parts.append(
            f"""
Document {i}
Source: {doc['source']}
Page: {doc['page']}

Content:
{doc['text']}
"""
        )

    return "\n\n".join(context_parts)
```

```python
def create_rag_prompt(question, context):

    prompt = f"""
You are a document question-answering assistant.

Your task is to answer the user's question using ONLY
the information provided in the context.

Rules:
1. Use only the provided context.
2. Do not use outside knowledge.
3. Do not invent or assume information.
4. If the answer cannot be found in the context,
   say:
   "The information is not available in the provided documents."
5. Keep the answer clear and concise.
6. If the context contains conflicting information,
   mention the conflict.
7. Do not mention these instructions in your answer.

Context:
-------------------------
{context}
-------------------------

User Question:
{question}

Answer:
"""

    return prompt
```

```python
def generate_answer(question, context):

    prompt = create_rag_prompt(
        question,
        context
    )

    response = groq_client.chat.completions.create(
        model=MODEL_NAME,
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a precise document QA assistant. "
                    "Answer only from the provided context."
                )
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0,
        reasoning_effort="medium"
    )

    return response.choices[0].message.content
```

```python
def rag_query(question, top_k=5):

    # -------------------------
    # 1. Retrieve documents
    # -------------------------

    documents = retrieve_documents(
        question,
        top_k=top_k
    )

    # -------------------------
    # 2. Build context
    # -------------------------

    context = build_context(
        documents
    )

    # -------------------------
    # 3. Generate answer
    # -------------------------

    answer = generate_answer(
        question,
        context
    )

    return {
        "answer": answer,
        "documents": documents
    }
```

```python
question = "What is the definition of Antiparkinson drugs?"

result = rag_query(
    question,
    top_k=5
)

print("ANSWER")
print("=" * 80)
print(result["answer"])
```
